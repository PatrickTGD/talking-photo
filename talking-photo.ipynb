# Talking Photo — free on Google Colab

Make one photo speak, using the free GPU.

**How to use:** press the **play button** on each box below, from top to bottom. Wait for each box to finish before the next one.

First: make sure the GPU is on — menu **Runtime → Change runtime type → T4 GPU → Save**.

In step 5 the page asks you for **one photo** (a clear photo of a face, looking at the camera, works best). The voice is already inside — Moss made it for you.

## Step 1 — Check the GPU

```
!nvidia-smi -L
```

## Step 2 — Get the program (SadTalker, free and open source). This takes a few minutes.

```
!git clone -q https://github.com/OpenTalker/SadTalker.git
%cd SadTalker
!pip install -q -r requirements.txt
!pip install -q "numpy==1.26.4"
```

## Step 3 — Download the AI model files (about 1.3 GB, a few minutes)

```
!mkdir -p checkpoints gfpgan/weights
B = "https://github.com/OpenTalker/SadTalker/releases/download/v0.0.2-rc"
!wget -q -nc $B/SadTalker_V0.0.2_256.safetensors -O checkpoints/SadTalker_V0.0.2_256.safetensors
!wget -q -nc $B/SadTalker_V0.0.2_512.safetensors -O checkpoints/SadTalker_V0.0.2_512.safetensors
!wget -q -nc $B/mapping_00109-model.pth.tar -O checkpoints/mapping_00109-model.pth.tar
!wget -q -nc $B/mapping_00229-model.pth.tar -O checkpoints/mapping_00229-model.pth.tar
!wget -q -nc https://github.com/TencentARC/GFPGAN/releases/download/v1.3.0/GFPGANv1.4.pth -O gfpgan/weights/GFPGANv1.4.pth
!ls -la checkpoints
```

## Step 4 — Get the voice, then upload YOUR photo when the page asks

```
!wget -q https://raw.githubusercontent.com/PatrickTGD/talking-photo/main/voice.mp3 -O /content/voice.mp3
from google.colab import files
print('Choose one clear face photo now:')
up = files.upload()
photo = list(up.keys())[0]
print('Photo ready:', photo)
```

## Step 5 — Make the video (about 1–3 minutes), then watch it and download it

```
import glob
!python inference.py --driven_audio /content/voice.mp3 --source_image "/content/{photo}" --checkpoint_dir checkpoints --result_dir /content/results --still --preprocess full --size 256
vids = sorted(glob.glob('/content/results/**/*.mp4', recursive=True))
print('Videos made:', vids)
from IPython.display import Video, display
display(Video(vids[-1], embed=True, width=480))
files.download(vids[-1])
```
